## Machine Learning

### Logistic Regression

Since the target column of the breast cancer data is categorical this machine learning problem is a classification problem. To start 
our model development we will use a logistic regression model which is good for basic classification tasks. Depending on the performance of the model we will decide to develop more powerful models to obtain better results

```python
# We will begin by importing the Logistic Regression model from the scikit-learn library
from sklearn.linear_model import LogisticRegression
```

```python
# Importing train test split from scikit-learn library
from sklearn.model_selection import train_test_split
```

```python
# Separating the data into features and target
X = data.iloc[:,2:]#feautures
y = data.iloc[:,1]#target
```

```python
# Splitting features and target into training and testing data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)
```

Since the values in the columns are of different scales of avrying magnitude, this can affect the quality of the model. To overcome this we will scale the columns to the same scale in order to enhance training of the model and accuracy of
the predictions. We will make use of the StandardScaler from the scikit-learn library

```python
# Importing StandarScaler
from sklearn.preprocessing import StandardScaler
```

```python
# Instanting StandardScaler as scaler
scaler = StandardScaler()
```

```python
# Instantiating a Logistic Regression model as logmodel
logmodel = LogisticRegression()
```

Instead of separately scaling the data and then training the model and predicting we can simplify this by
using a pipeline. We'll make use of the `make_pipeline` function from scikit-learn

```python
# Importing make_pipeline
from sklearn.pipeline import make_pipeline
```

```python
# Making a pipeline consisting of StandardScaler and the Logistic Regression model
pipeline_log = make_pipeline(scaler, logmodel)
```

```python
# Fitting the pipeline object to the training data 
pipeline_log.fit(X_train, y_train)
```

Now that we have trained the Logistic Regression model we can now predict on 
the test data and evaluate our model's performance

```python
# Making predictions on the test data
y_pred = pipeline.predict(X_test)
# Predicting probabilities to be used in roc_auc_score
y_pred_probs = pipeline.predict_proba(X_test)[:, 1]
```

```python
# Importing necessary tools to evaluate the model
from sklearn.metrics import confusion_matrix, classification_report, roc_auc_score
```

```python
# Printing the confusion matrix
print('For the Logistic Regression Model:\n')
print(confusion_matrix(y_test, y_pred))
print('\n')
# Printing the classification report
print(classification_report(y_test, y_pred))
```

It would appear that the Logistic Regression Model performed very well on this data set.

### Decision Tree Model

Let us now see whether a decision tree model can perform as well or even outperform the 
Logistic Regression model.

```python
# Importing the Decision Tree Classifier from scikit-learn library
from sklearn.tree import DecisionTreeClassifier
```

```python
# Creating a seed to use for random_state
SEED=101
```

```python
# Instantiating DecisionTreeClassifer as dtree
dtree = DecisionTreeClassifier(max_depth=3, random_state=SEED)
```

```python
# Creating a pipeline consisting of StandardScaler and DecisionTreeClassifier
pipeline_dtree = make_pipeline(scaler, dtree)
```

```python
# Fitting the pipeline to the data
pipeline_dtree.fit(X_train, y_train)
```

Now that we have trained our Decision Tree Model on the data we can now 
make predictions and evaluate the model's performance

```python
# Making predictions on the test_data
y_pred = pipeline_dtree.predict(X_test)
# Predicting probabilities to be used in roc_auc_score
y_pred_probs = pipeline_dtree.predict_proba(X_test)[:,1]
```

```python
# Printing the confusion matrix
print('For the Decision Tree Model:\n')
print(confusion_matrix(y_test, y_pred))
print('\n')
# Printing the classification report
print(classification_report(y_test, y_pred))
```

```python
# Printing the roc_auc_score
print(f'For the Decision Tree Model:\nROC_AUC_Score: {roc_auc_score(y_test, y_pred_probs):.2f}%')
```

The Decision Tree Model's results are also good. However it falls slightly below that of the Logistic Regression model. Let's finally try a Random Forest Model and see what the performance will be this time around

### Random Forests Model

Random Forests is an ensemble of multiple decision trees and so it should rightfully outperform
a single decision tree. However will it outperform the Logistic Regression model which has become our benchmark? Well let's see as we train, predict and finally evaluate our model

```python
# Importing RandomForestsClassifier from scikit-learn library
from sklearn.ensemble import RandomForestClassifier
```

```python
# Instantiating RandomForestsClassifier as rfc
rfc = RandomForestClassifier(n_estimators=25, random_state=SEED)
```

```python
# Making a pipeline consisting of StandardScaler and RandomClassifier
pipeline_rfc = make_pipeline(scaler, rfc)
```

```python
# Fitting the pipeline to the training data
pipeline_rfc.fit(X_train, y_train)
```

```python
# Predicting on the test data
y_pred = pipeline_rfc.predict(X_test)
# Predicting probabilites to be used in roc_auc_score
y_pred_probs = pipeline_rfc.predict_proba(X_test)[:,1]
```

```python
# Printing the confusion matrix
print('For the Random Forests Model:\n')
print(confusion_matrix(y_test, y_pred))
print('\n')
# Printing the classification report
print(classification_report(y_test, y_pred))
```

```python
# Printing the roc_auc_score
print(f'For the Random Forests Model:\nROC_AUC_Score: {roc_auc_score(y_test, y_pred_probs):.2f}%')
```

On average the Random Forests Model only did slighty better than the Decision Tree Model looking at the classification report and bit better on the roc auc score.\
However when comparing the confusion matrix and classification report of the Random Forests
Classifier with that of the Logistic Regression model it can be seen that it fell behind in both precision and recall.\
It is worth noting that since this is a breast cancer data set the target is diagnosing a 
tumor as either benign or malignant. This means that the metric of high importance in 
evaluating a model is **recall** as it is more important to reduce the rate of false negatives.\
In essence it would be much more costly to misdiagnose a malignant tumor as benign, than to 
misdiagnose a benign tumor as malignant.\
Therefore in this project a model's true value is in how high its **recall** metric is and currently the highest recall is the Logistic Regression model. Let's see whether another model can give us similar or even better results

### XGBoost

Extreme gradient boosting with the use of XGBoost is an effective machine learning
technique that makes use of powerful XGBoost models. Let's see whether this powerful model 
will be able to perform as well as the Logistic Regression model or even better.

```python
# Importing the xgboost library
import xgboost as xgb
```

```python
# Instanting XGBoostClassifier as xg_cl
xg_cl = xgb.XGBClassifier(objective='binary:logistic', seed=SEED, max_depth=3)
```

Unlike the previous machine learning models we built, XGBoost requires the target 
column be one hot encoded into 0s and 1s and not objects.\
In order to overcome this we will use LabelEncoder to convert the target column values
to the appropriate format to be used for training the XGBoostClassifier.

```python
# Importing LabelEncoder from scikit-learn library
from sklearn.preprocessing import LabelEncoder
```

```python
# Instantiating LabelEncoder as le
le = LabelEncoder()
```

```python
# Fitting and transforming the training target values
y_train_le = le.fit_transform(y_train)
```

```python
# Only transforming the test target values to avoid overfitting
y_test_le = le.transform(y_test)
```

```python
# Creating a pipeline consisting of StandardScaler and XGBoostClassifier
pipeline_xgcl = make_pipeline(scaler, xg_cl)
```

```python
# Fitting the pipeline object to the training data
pipeline_xgcl.fit(X_train, y_train_le)
```

```python
# Predicting on the test data
y_pred = pipeline_xgcl.predict(X_test)
# Predicting probabilities to use for roc_auc
y_pred_probs = pipeline_xgcl.predict_proba(X_test)[:,1]
```

```python
# Printing the confusion matrix
print('For the XGBoost Model:\n')
print(confusion_matrix(y_test_le, y_pred))
print('\n')
# Printing the classification report
print(classification_report(y_test_le, y_pred))
```

```python
# Printing the roc_auc_score
print(f'For the XGBoost Model:\nROC_AUC_Score: {roc_auc_score(y_test_le, y_pred_probs):.2f}%')
```

From the results it can be seen that the XGBoost model came the closest to the Logistic 
Regression model in terms of **recall** which is of importance.\
It is worth noting that the XGBoostClassifier here has not been tuned and so it is possible
that with proper hyperparamter tuning it may outperform the Logistic Regression model.\
Either ways it did quite well on the predictions so that is fine for now. Also this is a small
data set. Had it been a much larger data set the XGBClassifier may well outperform the Logistic 
Regression model
